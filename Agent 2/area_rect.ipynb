Write a graph that can calculate the Area of a rectangle.

input: {"length":90, "width":100}

output:"Area is: 9000"

```python
from typing import TypedDict
from langgraph.graph import StateGraph
```

```python
#State Schema

class AgentState(TypedDict):
    length:int
    width:int
    name:str
    results:str
    
```

```python
#create node

def get_area(state: AgentState)->AgentState:
    """Get the Area of the figure"""
    #update the state of the graph
    state['results']=f"The {state['name']} has an area of: {state['length']*state['width']} "

    return state


```

```python
#create the graph
graph=StateGraph(AgentState)

#add the node
graph.add_node('area', get_area)

#START the graph
graph.set_entry_point('area')

#END the graph
graph.set_finish_point('area')


#compile the graph
app=graph.compile()
```

```python
#Now lets visualize the flow of our graph
from IPython.display import Image, display
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
#now alter the state of the graph 
answer=app.invoke({'name':'Rectangle', 'length':89,'width':46})
```
